````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import beta
from sklearn.metrics import (accuracy_score, classification_report, confusion_matrix,
                             matthews_corrcoef, roc_auc_score, roc_curve)

pd.set_option("display.width", 160)

def wilson_interval(k, n, z=1.96):
    #95%-й интервал Уилсона для доли k/n.
    p = k / n
    centre = p + z * z / (2 * n)
    half = z * np.sqrt(p * (1 - p) / n + z * z / (4 * n * n))
    d = 1 + z * z / n
    return (centre - half) / d, (centre + half) / d

def credible_interval(k, n, level=0.95):
    #Байесовский интервал (как в Pocket AutoML): квантили Beta(k+1, n-k+1).
    a = (1 - level) / 2
    lo, hi = beta.ppf([a, 1 - a], k + 1, n - k + 1)
    return lo, hi

def per_class_table(cm, labels):
    #Метрики по матрице ошибок (строки — истина, столбцы — ответ модели).
    cm = np.asarray(cm, dtype=float)
    support = cm.sum(axis=1)
    recall = np.divide(np.diag(cm), support, out=np.zeros(len(labels)), where=support > 0)
    pred_cnt = cm.sum(axis=0)
    precision = np.divide(np.diag(cm), pred_cnt, out=np.zeros(len(labels)), where=pred_cnt > 0)
    f1 = np.divide(2 * precision * recall, precision + recall, out=np.zeros(len(labels)), where=(precision + recall) > 0)
    t = pd.DataFrame({"precision": precision, "recall": recall, "f1": f1, "support": support.astype(int)}, index=labels)
    t.loc["accuracy"] = [np.nan, np.nan, np.diag(cm).sum() / cm.sum(), int(cm.sum())]
    t.loc["macro avg"] = [precision.mean(), recall.mean(), f1.mean(), int(cm.sum())]
    w = support / support.sum()
    t.loc["weighted avg"] = [(precision * w).sum(), (recall * w).sum(), (f1 * w).sum(), int(cm.sum())]
    t["support"] = t["support"].astype(int)
    return t

def cm_to_labels(cm):
    #Матрица ошибок -> два списка (истина, ответ) — чтобы проверить расчёты функциями sklearn.
    yt, yp = [], []
    for i in range(len(cm)):
        for j in range(len(cm)):
            yt += [i] * int(cm[i][j]); yp += [j] * int(cm[i][j])
    return np.array(yt), np.array(yp)

def plot_cm(cm, labels, title, ax=None):
    cm = np.asarray(cm)
    ax = ax or plt.gca()
    ax.imshow(cm, cmap="Blues")
    ax.set_xticks(range(len(labels))); ax.set_xticklabels(labels, rotation=20)
    ax.set_yticks(range(len(labels))); ax.set_yticklabels(labels)
    ax.set_xlabel("Ответ модели"); ax.set_ylabel("Настоящий класс"); ax.set_title(title)
    for i in range(len(labels)):
        for j in range(len(labels)):
            ax.text(j, i, int(cm[i, j]), ha="center", va="center", fontsize=14,
                    color="white" if cm[i, j] > cm.max() / 2 else "black")

print("Функции готовы.")
````

````python
labels1 = ["хот-дог", "не хот-дог"]
cm_tm = np.array([[33, 5],
                  [4, 34]])          # строки — настоящий класс, столбцы — ответ модели (тестовые 76 фото)

print("Метрики на тестовой выборке Teachable Machine (76 фото):")
tbl = per_class_table(cm_tm, labels1)
display(tbl.round(3))

tp, fn, fp, tn = cm_tm[0, 0], cm_tm[0, 1], cm_tm[1, 0], cm_tm[1, 1]
yt, yp = cm_to_labels(cm_tm)
extra = pd.Series({
    "Accuracy": (tp + tn) / cm_tm.sum(),
    "Специфичность (доля верно отвергнутых «не хот-догов»)": tn / (tn + fp),
    "Balanced accuracy": (tp / (tp + fn) + tn / (tn + fp)) / 2,
    "MCC": matthews_corrcoef(yt, yp),
    "Ложных тревог (FP)": fp, "Пропусков (FN)": fn,
})
display(extra.round(3).to_frame("значение"))
lo, hi = wilson_interval(tp + tn, cm_tm.sum())
print(f"95%-й интервал Уилсона для accuracy: от {lo:.1%} до {hi:.1%} (n = {cm_tm.sum()})")

fig, ax = plt.subplots(figsize=(4.5, 4))
plot_cm(cm_tm, labels1, "Тест Teachable Machine (76 фото)", ax)
plt.tight_layout(); plt.show()
````

**Как читать результат.**
- **Accuracy = 88,2%** (67 из 76). Классы равные (по 38), поэтому accuracy честная.
- **Precision и recall почти одинаковы** (для хот-дога 0,89 и 0,87; для «не хот-дога» 0,87 и 0,89), то есть модель **не смещена** в одну сторону: пропуски (5) и ложные тревоги (4) встречаются примерно одинаково часто. F1 ≈ 0,88 для обоих классов.
- **MCC ≈ 0,76** — заметно выше нуля, это подтверждает, что результат сильно лучше случайного угадывания (у случайной модели было бы около 0).
- **Интервал Уилсона 79–94%**: тестовых фото всего 76, поэтому настоящая точность может быть в этих пределах. Чтобы проверить надёжнее, потребовалась внешняя проверка на 500 фото (ниже).

````python
import os
import numpy as np
import tensorflow as tf
from PIL import Image, ImageOps
from sklearn.metrics import classification_report, accuracy_score

model_path = "model.savedmodel"
model = tf.saved_model.load(model_path)
infer = model.signatures["serving_default"]
with open("labels.txt", "r", encoding="utf-8") as f:
    class_names = [line.strip() for line in f.readlines()]
IMAGE_DIR = "hot_dog_validation"  
exts = (".jpg", ".jpeg", ".png", ".bmp", ".webp")
HOT_DOG_IDX=0
y_true = [] 
y_pred = [] 
filenames = []

print("\nРезультаты по каждому файлу ")
for filename in os.listdir(IMAGE_DIR):
    if not filename.lower().endswith(exts):
        continue
    path = os.path.join(IMAGE_DIR, filename)
    try:
        image = Image.open(path).convert("RGB")
        image = ImageOps.fit(image, (224, 224), Image.Resampling.LANCZOS)
        image_array = np.asarray(image).astype(np.float32)
        normalized = (image_array / 127.5) - 1
        data = np.expand_dims(normalized, axis=0)

        prediction = infer(tf.constant(data))
        output_key = list(prediction.keys())[0]
        output_data = prediction[output_key].numpy()

        index = np.argmax(output_data)
        confidence = output_data[0][index]

        y_true.append(HOT_DOG_IDX)
        y_pred.append(index)
        filenames.append(filename)

        is_correct = "+" if index == HOT_DOG_IDX else "-"
        class_name = class_names[index].strip()
        print(f"{is_correct} | {filename}: предсказан {class_name[2:]} ({confidence:.2%})")

    except Exception as e:
        print(f"{e}")

print("\n ИТОГОВЫЕ МЕТРИКИ")

if len(y_true) == 0:
    print("Нет данных для подсчета метрик. Проверь папку с картинками.")
else:
    acc = accuracy_score(y_true, y_pred)
    print(f"Accuracy (Доля верных ответов): {acc:.2%}\n")
    clean_names = [name[2:] if len(name) > 2 else name for name in class_names]
    
    report = classification_report(
        y_true, 
        y_pred, 
        labels=[0, 1], 
        target_names=clean_names[:2], 
        zero_division=0
    )
    print("Детальный отчет по классам:")
    print(report)
````

Модель правильно распознала 86% фотографий (215 из 250). Остальные 14% (35 картинок) она ошибочно назвала "не хот-догами".

**результат**

- Поэтому результат 100% нельзя считать доказательством идеальной модели, это лишь означает, что на маленьком тесте ошибок не оказалось. Чтобы проверить модель честно, нужны другие фото, которых она не видела (папка `mix`, ниже).
- Классы хорошо различимы визуально (горы — вертикальные силуэты и снег, море — ровный горизонт и синие тона, водопад — вода среди зелени и скал), поэтому такая небольшая модель справилась. Ошибки следует ждать на «пограничных» фото: водопад в горах, горное озеро, морская волна на скалах.

````python
import os
import numpy as np
import tensorflow as tf
from PIL import Image, ImageOps

model_path = "model.savedmodel2"
model = tf.saved_model.load(model_path)
infer = model.signatures["serving_default"]

with open("labels2.txt", "r", encoding="utf-8") as f:
    class_names = [line.strip() for line in f.readlines()]
clean_names = [name[2:].strip() if len(name) > 2 else name.strip() for name in class_names]

IMAGE_DIR = "mix" 
exts = (".jpg", ".jpeg", ".png", ".bmp", ".webp")

predictions = []
confidences = []
print("Результаты по файлам:")

for filename in os.listdir(IMAGE_DIR):
    if not filename.lower().endswith(exts):
        continue
        
    path = os.path.join(IMAGE_DIR, filename)
    try:
        image = Image.open(path).convert("RGB")
        image = ImageOps.fit(image, (224, 224), Image.Resampling.LANCZOS)
        image_array = np.asarray(image).astype(np.float32)
        normalized = (image_array / 127.5) - 1
        data = np.expand_dims(normalized, axis=0)

        prediction = infer(tf.constant(data))
        output_key = list(prediction.keys())[0]
        output_data = prediction[output_key].numpy()

        index = np.argmax(output_data)
        confidence = output_data[0][index]
        
        predictions.append(index)
        confidences.append(confidence)
        
        print(f"{filename} -> {clean_names[index]} ({confidence:.2%})")
    except Exception as e:
        print(f"Ошибка с {filename}: {e}")

print("ОЦЕНКА УВЕРЕННОСТИ МОДЕЛИ")

if len(predictions) == 0:
    print("Нет картинок для анализа.")
else:
    total = len(predictions)
    print(f"Всего проверено картинок: {total}\n")
    
    print("Распределение по классам и средняя уверенность:")
    for i, name in enumerate(clean_names):
        count = predictions.count(i)
        if count > 0:
            avg_conf = np.mean([confidences[j] for j in range(total) if predictions[j] == i])
            print(f" - {name}: {count} шт. (средняя уверенность: {avg_conf:.2%})")
        else:
            print(f" - {name}: 0 шт.")
            
    print("\nКачество распознавания (по уверенности):")
    high_conf = sum(1 for c in confidences if c >= 0.90)
    mid_conf = sum(1 for c in confidences if 0.60 <= c < 0.90)
    low_conf = sum(1 for c in confidences if c < 0.60)
    
    print(f" - Высокая уверенность (>90%): {high_conf} картинок ({high_conf/total:.1%})")
    print(f" - Средняя уверенность (60-90%): {mid_conf} картинок ({mid_conf/total:.1%})")
    print(f" - Низкая уверенность (<60%): {low_conf} картинок ({low_conf/total:.1%})")
````

## 4 задание

### Подробное описание API и приложения (задание 4)

**Модель.** В API подключена модель, обученная в Pocket AutoML (задание 3) и экспортированная в формате **TensorFlow Lite**.

**Архитектура.**
```
 Приложение (Tkinter)  --HTTP-->  API (FastAPI)  -->  модель TensorFlow Lite
 выбор файла, показ результата     POST /predict        предобработка → нейросеть → вероятности
```
Приложение само модель не запускает: оно только отправляет файл в API и показывает ответ.

**Эндпоинты API (FastAPI)**

| Метод | Адрес | Назначение |
|---|---|---|
| POST | `/predict` | принимает изображение (поле `file`), возвращает класс, точность и вероятности всех классов |
| GET | `/model/info` | классы, размер входа модели, чем выполняется модель |
| GET | `/health` | проверка работы сервера |
| GET | `/docs` | интерактивная документация (Swagger), в ней запросы можно пробовать в браузере |

**Что возвращает `/predict`:**

| Поле | Смысл |
|---|---|
| `label` | результат распознавания — самый вероятный класс (earth / sun / moon) |
| `confidence`, `confidence_percent` | **«точность распознавания»** — уверенность модели в этом ответе (вероятность класса, 0–1 и в %) |
| `probabilities` | вероятности всех трёх классов по убыванию (в сумме 100%) |
| `low_confidence` | `true`, если уверенность ниже порога 60%: ответу стоит доверять осторожно |
| `inference_ms`, `backend` | время работы модели и чем она выполнена |

**Ошибки API:** 400 — файл пустой или не является изображением; 413 — файл больше 10 МБ.

**Приложение.** Кнопка «Выбрать изображение…» открывает файл, показывает превью и сразу отправляет его в API. Справа выводятся: результат (по-русски — Земля, Луна, Солнце), «Точность распознавания» в процентах с полосой, вероятности по каждому из трёх классов, время работы и предупреждение при низкой уверенности. Если сервер недоступен, пользователь видит понятное сообщение. Запросы выполняются в отдельном потоке, окно не зависает.

**Как запустить:** `uvicorn api.main:app --reload` (сервер) и `python -m app.main` (приложение).

**Общие наблюдения.**
1. С небольшим числом фото (20–40 на класс) Teachable Machine и Pocket AutoML дают хорошие результаты на «типичных» изображениях
2. Чем больше проверочных фото, тем честнее оценка: 100% на 10 фото (задание 2) и 86,8% на 500 фото (задание 1) нельзя сравнивать напрямую — у первого интервал огромный.
3. Accuracy нужно смотреть вместе с precision, recall и матрицей ошибок: они показывают, какого рода ошибки делает модель (ложные тревоги или пропуски). Метрики по одной папке с фото единственного класса (как в первых выводах кода) вводят в заблуждение: нули и «1.00» появляются из-за пустых классов.
4. Уверенность модели и точность — разные вещи: уверенность относится к одному фото и может быть высокой при ошибке, точность — ко всей выборке.
5. Основные причины ошибок: малое число обучающих фото, похожие классы, нетипичные ракурсы и фон, лишние объекты на фото.
